##### Chunking

```python
documents = [
    {
        "id": "return-policy",
        "title": "Shoply Return Policy",
        "source": "return-policy.pdf",
        "category": "return",
        "text": """
        Shoply Return Policy

        Customers can return unused products within 30 days of purchase.
        The product must be in its original packaging and must not show
        signs of damage or use.

        Customers should contact Shoply customer support before sending
        the product back. The support team will provide return instructions.

        Some products are not eligible for return. Personalized products,
        digital products, and clearance items cannot normally be returned.

        Once Shoply receives the returned product, the product is inspected.
        If the return is approved, the refund is processed according to the
        refund policy.
        """
    }
]
```

```python
document = documents[0]
print(document)
type(document)
```

```python
print("Document ID:", document["id"])
print("Title:", document["title"])
print("Source:", document["source"])
print("Category:", document["category"])

print("\nDocument:")
print(document["text"])
```

```python
def split_into_paragraphs(text):
  paragraphs = text.split("\n\n")
  return [type(paragraphs), paragraphs]

para_type, paragraphs = split_into_paragraphs(document["text"])
print(para_type)
print(paragraphs)
```

```python
def split_into_paragraphs(text):
  paragraphs = text.split("\n\n")
  
  paragraphs = [
		paragraph.strip() for paragraph in paragraphs if paragraph.strip()
	] 
  
  return paragraphs

result = split_into_paragraphs(document["text"])
print(result)
```

```python
text = document["text"]
paragraphs = split_into_paragraphs(text)
print("Number of paragraphs: ", len(paragraphs))

for i, paragraph in enumerate(paragraphs):
  print(f"\nParagraph {i}:")
  print(paragraph)
```

##### Word-based chunking

```python
def chunk_text(text, chunk_size=40):
  words = text.split()
  
  chunks = []
  
  for start in range(0, len(words), chunk_size):
    end = start + chunk_size	
    chunk = " ".join(words[start:end])
    chunks.append(chunk)
  
  return chunks


chunks = chunk_text(document["text"], chunk_size=40)
print("Number of chunks:", len(chunks))

for chunk in chunks:
  print(f"\n{chunk}")
```
