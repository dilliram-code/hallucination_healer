##### Word-based chunking

```python
def chunk_text(text, chunk_size=40):
  words = text.split()
  
  chunks = []
  
  for start in range(0, len(words), chunk_size):
    end = start + chunk_size	
    chunk = " ".join(words[start:end])
    chunks.append(chunk)
  
  return chunks


chunks = chunk_text(document["text"], chunk_size=40)
print("Number of chunks:", len(chunks))

for chunk in chunks:
  print(f"\n{chunk}")
```

##### Chunk overlap implementation

```python
def chunk_with_overlap(text, chunk_size=40, overlap=10):
  words = text.split()
  chunks = []
  start = 0
  
  while start < len(words):
    end = start + chunk_size
    chunk = " ".join(words[start: end])
    chunks.append(chunk)
    start = end - overlap
  
  return chunks

chunks = chunk_with_overlap(text)
for i, chunk in enumerate(chunks):
  print(f"\nChunk {i+1}:\n{chunk}")
```

##### chunking: overlap based splitting

```python
def create_chunks(document, chunk_size=40, overlap=10):
  words = document["text"].split()
  chunks = []
  start = 0
  chunk_number = 0
  
  while start < len(words):
    end = start + chunk_size
    text = " ".join(words[start:end])
    
    chunk = {
            "chunk_id": f"{document['id']}_chunk_{chunk_number}",
            "document_id": document["id"],
            "text": text,
            "source": document["source"],
            "title": document["title"],
            "category": document["category"],
            "chunk_index": chunk_number,
        }
    
    chunks.append(chunk)
    
    chunk_number += 1
    start = end - overlap
  
  return chunks

def create_all_chunks(documents):
  """Create chunks for all documents"""
  all_chunks = []
  for document in documents:
    chunks = create_chunks(document)
```

```python
import math


# ============================================================
# 1. DUMMY DOCUMENTS
# ============================================================

documents = [

    {
        "id": "return-policy",
        "source": "return-policy.pdf",
        "category": "return",
        "text": """
        Customers can return unused products within 30 days of purchase.
        The product must be in its original packaging.
        """
    },

    {
        "id": "shipping-policy",
        "source": "shipping-policy.pdf",
        "category": "shipping",
        "text": """
        Standard shipping usually takes 3 to 5 business days.
        Express shipping usually takes 1 to 2 business days.
        """
    },

    {
        "id": "refund-policy",
        "source": "refund-policy.pdf",
        "category": "refund",
        "text": """
        Approved refunds are processed within 7 business days.
        Refunds are normally sent to the original payment method.
        """
    },

    {
        "id": "warranty-policy",
        "source": "warranty-policy.pdf",
        "category": "warranty",
        "text": """
        Electronic products are covered by a one-year warranty.
        The warranty covers manufacturing defects.
        """
    },
]


# ============================================================
# 2. CHUNKING
# ============================================================

def create_chunks(
    document,
    chunk_size=20,
    overlap=5,
):

    words = document["text"].split()

    chunks = []

    start = 0
    chunk_number = 0

    while start < len(words):

        end = start + chunk_size

        text = " ".join(words[start:end])

        chunks.append(
            {
                "chunk_id": (
                    f"{document['id']}_chunk_{chunk_number}"
                ),
                "text": text,
                "source": document["source"],
                "category": document["category"],
            }
        )

        chunk_number += 1

        start = end - overlap

    return chunks


# ============================================================
# 3. CREATE ALL CHUNKS
# ============================================================

all_chunks = []

for document in documents:

    chunks = create_chunks(document)

    all_chunks.extend(chunks)


# ============================================================
# 4. DUMMY EMBEDDING MODEL
# ============================================================

CONCEPTS = {

    "return": [
        "return",
        "returned",
        "back",
    ],

    "shipping": [
        "shipping",
        "delivery",
        "shipped",
        "tracking",
    ],

    "refund": [
        "refund",
        "money",
        "payment",
    ],

    "warranty": [
        "warranty",
        "defect",
    ],

    "time": [
        "day",
        "days",
        "week",
        "weeks",
        "year",
        "years",
    ],
}


def embed_text(text):

    text = text.lower()

    vector = []

    for keywords in CONCEPTS.values():

        score = 0

        for keyword in keywords:

            if keyword in text:
                score += 1

        vector.append(float(score))

    return normalize(vector)


# ============================================================
# 5. NORMALIZATION
# ============================================================

def normalize(vector):

    magnitude = math.sqrt(
        sum(value ** 2 for value in vector)
    )

    if magnitude == 0:
        return vector

    return [
        value / magnitude
        for value in vector
    ]


# ============================================================
# 6. CREATE EMBEDDINGS FOR CHUNKS
# ============================================================

for chunk in all_chunks:

    chunk["embedding"] = embed_text(
        chunk["text"]
    )


# ============================================================
# 7. COSINE SIMILARITY
# ============================================================

def cosine_similarity(a, b):

    dot_product = sum(
        x * y
        for x, y in zip(a, b)
    )

    magnitude_a = math.sqrt(
        sum(x ** 2 for x in a)
    )

    magnitude_b = math.sqrt(
        sum(y ** 2 for y in b)
    )

    if magnitude_a == 0 or magnitude_b == 0:
        return 0.0

    return (
        dot_product
        / (magnitude_a * magnitude_b)
    )


# ============================================================
# 8. RETRIEVAL
# ============================================================

def retrieve(
    query,
    chunks,
    top_k=3,
):

    query_vector = embed_text(query)

    results = []

    for chunk in chunks:

        score = cosine_similarity(
            query_vector,
            chunk["embedding"],
        )

        results.append(
            {
                "chunk": chunk,
                "score": score,
            }
        )

    results.sort(
        key=lambda item: item["score"],
        reverse=True,
    )

    return results[:top_k]


# ============================================================
# 9. USER QUERY
# ============================================================

query = "How long do I have to return a product?"


# ============================================================
# 10. RETRIEVE
# ============================================================

results = retrieve(
    query,
    all_chunks,
    top_k=3,
)


# ============================================================
# 11. DISPLAY
# ============================================================

print("\nUSER QUESTION")
print(query)

print("\nQUERY VECTOR")
print(embed_text(query))

print("\nRETRIEVED RESULTS")

for rank, result in enumerate(
    results,
    start=1,
):

    chunk = result["chunk"]

    print("\n---------------------------")

    print("Rank:", rank)

    print(
        "Similarity:",
        round(result["score"], 4)
    )

    print(
        "Chunk:",
        chunk["chunk_id"]
    )

    print(
        "Source:",
        chunk["source"]
    )

    print(
        "Category:",
        chunk["category"]
    )

    print(
        "Text:",
        chunk["text"]
    )
```
